Dataset: AB_NYC_2019 (48,895 listings, 16 columns)
Goal: find and fix missing values, duplicates, wrong data types and inconsistent values.

```python
before = {
    "rows": df_raw.shape[0],
    "columns": df_raw.shape[1],
    "missing_total": int(df_raw.isnull().sum().sum()),
    "duplicates": int(df_raw.duplicated().sum()),
}
before
```

```python
df['reviews_per_month'] = df['reviews_per_month'].fillna(0)
```

```python
df['name'] = df['name'].fillna("Unknown")
df['host_name'] = df['host_name'].fillna("Unknown")
```

Filled reviews_per_month with 0 for listings with no reviews (not truly missing — 0 reviews means 0 per month). Filled name and host_name with "Unknown" (37 rows total). Left last_review blank for listings with no reviews, since there is no valid date to assign.

```python
df.isnull().sum()
```

```python
dupes = df[df.duplicated(subset=['host_id', 'name'], keep=False)]
dupes.sort_values(['host_id', 'name']).head(20)
```

```python
near_dupes_count = df.duplicated(subset=['host_id', 'name'], keep=False).sum()
print(f"Found {near_dupes_count} rows sharing host_id + name — inspected and confirmed these are distinct units (different id, lat/long, review history), not duplicates. No rows dropped.")
```

Checked for duplicates two ways: (1) exact duplicate rows → 0 found. (2) Same host_id + name → 243 rows found, but inspection showed each has a unique id, different coordinates, and different review counts — these are separate rooms with reused listing names (e.g., hotels with many similarly-named units), not duplicates. No rows were removed at this stage.

```python
df['last_review'] = pd.to_datetime(df['last_review'])
df['neighbourhood_group'] = df['neighbourhood_group'].astype('category')
df['neighbourhood'] = df['neighbourhood'].astype('category')
df['room_type'] = df['room_type'].astype('category')

df.info()
```

```python
df['name'] = df['name'].str.strip()
```

```python
(df['name'] != df['name'].str.strip()).sum()
```

Removed extra spaces from the start and end of listing names (238 rows affected). Left the capitalization style as-is since it reflects intentional host choices, not a data error.

```python
df = df[df['price'] > 0]
```

```python
df = df[df['minimum_nights'] <= 365]
```

```python
df.shape
```

Removed 11 listings with price == 0 (invalid — Airbnb listings cannot be free) and 14 listings with minimum_nights over 365 (unrealistic booking requirement). Rows dropped from 48,895 to 48,870.

```python
df.to_csv("data/AB_NYC_2019_cleaned.csv", index=False)
```
